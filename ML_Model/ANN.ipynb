```python
# ---- Dataset: Student Performance (Set A) ----
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n_samples = 300

study_hours = rng.uniform(0, 20, n_samples)               # hours/week
attendance = rng.uniform(40, 100, n_samples)               # percent
assignments = rng.integers(0, 11, n_samples)                # count out of 10

# Underlying rule (kept secret from students) + noise, to create a learnable but non-trivial pattern
score = 0.25 * study_hours + 0.06 * attendance + 0.8 * assignments + rng.normal(0, 3, n_samples)
label = (score > np.median(score)).astype(int)              # 1 = Pass, 0 = Fail

df = pd.DataFrame({
    "study_hours_per_week": study_hours,
    "attendance_percent": attendance,
    "assignments_completed": assignments,
    "result": label
})
df.to_csv("student_performance_setA.csv", index=False)
df.head()
```
