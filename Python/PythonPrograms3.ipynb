```python
def square(n):
  return n*n
result=square(5)
print("Output:",result)
def cube(n):
  return n*n*n
result=cube(5)
print("Output:",result)
```

```python
def check_even_odd(n):
  if n%2==0:
    return "Even"
  else:
    return "Odd"
check_even_odd(5)
```

```python
def check_number(n):
  if n>=0:
    print(n,"is positive")
  elif n<=0:
    print(n,"is negative")
  else:
    print(n,"is zero")
check_number(-3)
```

```python
def maximumOfTwo(a,b):
  if a>b:
    print(a,"is greater than",b)
  else:
    print(b,"is greater than",a)
a=int(input("Enter your number:"))
b=int(input("Enter your number:"))
maximumOfTwo(a,b)
```

```python
def minimumOfThree(a, b, c):
    if a <= b and a <= c:
        return a
    elif b <= a and b <= c:
        return b
    else:
        return c
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

print(minimumOfThree(a, b, c))
```

```python
#Level-2(Functions and conditions)
def is_eligible(age):
  if age>=18:
    return "eligible to vote"
  else:
    return "not eligible to vote"
is_eligible(1)
```

```python
#leap_year
def is_leap_year(year):
  if year%4==0:
    if year%100==0:
      if year%400==0:
        return "Leap Year"
      else:
        return "Not a Leap year"
    else:
      return "Leap Year"
  else:
    return "Not a Leap year"
is_leap_year(2000)
```

```python
#Grade Claculator
def calculate_grade(marks):
  if marks>=90:
    return "A"
  elif marks>=80:
    return "B"
  elif marks>=70:
    return "C"
  elif marks>=60:
    return "D"
  else:
    return "F"
marks=int(input("Enter your marks:"))
calculate_grade(marks)
```

```python
#Divisibility
def is_divisible(n):
  if n%5==0 and n%11==0:
    return "Divisible"
  else:
    return "Not Divisible"
is_divisible(110)
```
