```python
#Take a number from the user and print the sum of all numbers from 1 to that number.
num=int(input("Enter Your number:"))
sum=0
for i in range(1,num+1):
  sum+=i
print("Output:",sum)
```

```python
#Take a number and print its multiplication table from 1 to 10.
num=int(input("Enter Your Number:"))
for i in range(1,11):
  print(num,"*",i,"=",num*i)
```

```python
#Take a number and find how many digits it has
num=int(input("Enter Your Number:"))
count=0
while num>0:
  num=num//10
  count+=1
print("Output:",count)
```

```python
#Take a number and find the sum of its digits.
num=int(input("Enter Your Number:"))
sum=0
while num>0:
  digit=num%10
  sum+=digit
  num=num//10

print("Sum of digits:",sum)
```

```python
#Take a string and determine whether it is a palindrome.
string=input("Enter Your String:")
if(string==string[::-1]):
  print("Given String is Palindrome")
else:
  print("Given String is not palindrome")
```

```python
#Take a number from the user and determine whether it is prime.
num=int(input("Enter Your Number:"))
for i in range(2,num):
  if num%i==0:
    print(num,"is not prime")
    break
else:
  print(num,"is prime")
```

```python
#reverse of a number usig while
num=int(input("Enet your number:"))
reverse=0
while num>0:
  digit=num%10
  reverse=reverse*10+digit
  num=num//10
print("Reverse:",reverse)
```

```python
#factorial
num=int(input("Enter your number:"))
factorial=1
while num>0:
  factorial=factorial*num
  num=num-1
print(factorial)
```

```python
#Palindrome of number
num=int(input("Enter your number:"))
reverse=0
original=num
while num>0:
  digit=num%10
  reverse=reverse*10+digit
  num=num//10
if(original==reverse):
  print("Palindrome")
else:
  print("Not a palindrome")
```

```python
#ArmStrong Number
num=int(input("Enter your number:"))
original=num
total=0
while num>0:
  digit=num%10
  total=total+digit**3
  num=num//10
print("Armstrong Number:",total)
if(original==total):
  print("Armstrong Number")
else:
  print("Not a Armstrong Number")
```

```python
#Fibonacci series
num=int(input("Enter your number:"))
first=0
second=1
print(first)
print(second)
for i in range(2,num):
  third=first+second
  print(third)
  first=second
  second=third
```
