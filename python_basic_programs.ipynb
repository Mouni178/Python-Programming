```python
#Variables + input + type conversion-Take the user's name, age, and CGPA as input and print them in a meaningful sentence.
name=input("Enter name:")
age=int(input("Enter Age:"))
cgpa=float(input("Enter cgpa:"))
print("My name is",name)
print("I am",age,"years old")
print("My cgpa is ",cgpa)
```

```python
#Operators-take two numbers and print add,sub,mul,div,remainder
num1=float(input("Enter first number:"))
num2=float(input("Enter second number:"))
add=num1+num2
sub=num1-num2
mul=num1*num2
div=num1/num2
rem=num1%num2
print("Addition:",add)
print("Subtraction:",sub)
print("Multiplication:",mul)
print("Division:",div)
print("Remainder:",rem)
```

```python
#Conditional Statements-Take a number and print whether it is positive,negative or zero
num=int(input("Enter your number:"))
if num>0:
  print(num,"is Positive number")
elif num<0:
  print(num,"is negative number")
else:
  print(num,"is zero")
```

```python
#Even-odd
num=int(input("Enter Your number:"))
if num%2==0:
  print(num,"is Even Number")
else:
  print(num,"is Odd Number")
```

```python
#Largest number-Take three numbers and print the largest.
num1=float(input("Enter first number:"))
num2=float(input("Enter second number:"))
num3=float(input("Enter third number:"))
if num1>num2 and num1>num3:
  print(num1,"is largest number")
elif num2>num1 and num2>num3:
  print(num2,"is largest number")
else:
  print(num3,"is largest number")
```

```python
#Loops-Print all numbers from 1 to 20 that are divisible by 3
n=20
for i in range(1,n+1):
  if i%3==0:
    print(i)
```

```python
#Loop + counting-Take a number n and count how many even numbers exist from 1 to n
num=int(input("Enter Number:"))
count=0
for i in range(1,num+1):
  if i%2==0:
    count+=1
    print(i)
print("Count of Even numbers:",count)
```

```python
#String Operations
string=input("Enter string:")
print("Length:",len(string))
print("First Character:",string[0])
print("Last Character:",string[-1])
print("Reverse:",string[::-1])
print("UpperCase:",string.upper())
print("LowerCase:",string.lower())
print("TitleCAse:",string.title())
print("Slicing:",string[2:5])
print("Repetition:",string*3)
print("SwapCase:",string.swapcase())
print("Searching:",string.find('y'))
print("Indexing:",string.index('o'))
print("Count:",string.count("m"))
print("StartsWith:",string.startswith('M'))
print("EndsWith:",string.endswith('a'))
print("Replacing:",string.replace('n','l'))
```

```python
#For Sentences
sentence=input("Enter the sentence:")
print("Splitting",sentence.split())
print("Number of words:",len(sentence.split()))
words=sentence.split()
print("Join Words:"," ".join(words))
print("Removing Extra Spaces:",sentence.strip())
print("Replacing Words:",sentence.replace("Python", "Java"))
print("Finding position of word:",sentence.find("Python"))
```
